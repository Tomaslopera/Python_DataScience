Desarrollar un dashboard interactivo con Plotly y Dash para visualizar los datos de terremotos proporcionados

1. Cargar y preparar los datos
2. Creación de gráficos
    - Mapa Coroplético (Densidad de terremotos por región)
    - Gráfico de Barras (Número de terremotos por magnitud)
    - Gráfico de Lineas (Tendencia de la magnitud de los terremotos a lo largo del tiempo)
    - Gráfico de Dispersión (Relación entre la magnitud de los terremotos)
    - Gráfico 3D (Relación entre magnitud, profundidad y tiempo)
3. Diseño de la aplicación (Utilizando Dash)

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import dash
from dash import dcc, html
```

### Cargar y preparar datos

```python
df = pd.read_csv('../Python para Data Science/Datasets/earthquake+data.csv')
df.head()
```

```python
df['Date & Time'] = pd.to_datetime(df['Date & Time'])
df.info()
```

### Mapa Coroplético

```python
# fig1 = px.choropleth(df, locations='Country', locationmode='country names', color='Magnitude', hover_name='Country', projection='natural earth')
# fig1.update_layout(title_text='Mapa Global de Terremotos según Magnitud')
```

### Gráfico de barras

```python
magnitudes = df['Magnitude'].value_counts().sort_index()
fig2 = go.Figure(data=[go.Bar(x=magnitudes.index, y=magnitudes.values
                    , marker = dict(
                        color = magnitudes.values,
                        colorscale = 'Viridis',
                        showscale = True,
                        colorbar = dict(title='Número de Terremotos')
                    ), text = magnitudes.values, textposition='auto')])
fig2.update_layout(title_text='Número de Terremotos por Magnitud', xaxis_title='Magnitud', yaxis_title='Número de Terremotos')
```

### Gráfico de lineas

```python
df.sort_values('Date & Time', inplace=True)
fig3 = go.Figure(go.Scatter(x=df['Date & Time'], y=df['Magnitude'], mode='lines', line=dict(color='blue', width=2), text=df['Magnitude'], textposition='top center'))
fig3.update_layout(title_text='Evolución de la Magnitud de los Terremotos', xaxis_title='Fecha', yaxis_title='Magnitud')
```

### Gráfico de dispersión

```python
fig4 = go.Figure(go.Scatter(x=df['Depth'], y=df['Magnitude'], mode='markers', marker=dict(size=10, color=df['Magnitude'], colorscale='Viridis', showscale=True, colorbar=dict(title='Magnitud')), text=df['Country'], hoverinfo='text+x+y'))
fig4.update_layout(title_text='Magnitud de los Terremotos según Profundidad', xaxis_title='Profundidad', yaxis_title='Magnitud')
```

### Gráfico 3D

```python
fig5 = go.Figure(go.Scatter3d(x=df['Date & Time'], y=df['Depth'], z=df['Magnitude'], mode='markers', marker=dict(size=5, color=df['Magnitude'], colorscale='Viridis', opacity=0.8)))
fig5.update_layout(title_text='Magnitud de los Terremotos según Profundidad y Fecha', scene=dict(xaxis_title='Fecha', yaxis_title='Profundidad', zaxis_title='Magnitud'))
```

### Creación Dashboard

```python
app = dash.Dash(__name__)
```

```python
app.layout = html.Div([
    html.H1('Dashboard de Terremotos', style={'text-align': 'center'}),
    # dcc.Graph(figure=fig1),
    dcc.Graph(figure=fig2),
    dcc.Graph(figure=fig3),
    dcc.Graph(figure=fig4),
    dcc.Graph(figure=fig5)
])
```
